# Advanced Matplotlib plotting

This tutorial covers horizontal bars, histograms, filled regions,
reference spans, arrows, infinitely extending lines, and secondary
y-axes.

```python
import numpy as np
from maxplotlib import Canvas

x = np.linspace(0, 2 * np.pi, 200)
```

## Common plotting primitives

```python
canvas, ax = Canvas.subplots(width="12cm", ratio=0.65)
ax.barh([0, 1, 2], [2, 4, 3], color="steelblue", alpha=0.8)
ax.set_yticks([0, 1, 2], labels=["A", "B", "C"])
ax.set_xlabel("Amount")
ax.set_title("Horizontal bars")
canvas.show()
```

```python
samples = np.random.default_rng(4).normal(size=1000)
canvas, ax = Canvas.subplots()
ax.hist(samples, bins=30, color="slateblue", alpha=0.75)
ax.set_xlabel("Value")
ax.set_ylabel("Count")
ax.set_title("Distribution")
canvas.show()
```

## Steps, stairs, broken bars, and pie charts

```python
canvas, ax = Canvas.subplots()
ax.step([0, 1, 2, 3], [1, 3, 2, 4], where="mid", label="step")
ax.stairs([1, 2, 1], edges=[0, 1, 2, 3], color="purple", label="stairs")
ax.set_title("Discrete data")
ax.set_legend(True)
canvas.show()
```

```python
canvas, ax = Canvas.subplots()
ax.broken_barh([(0, 1), (1.5, 0.75), (2.75, 1.0)], (0, 0.6), color="orange")
ax.set_xlabel("Intervals")
canvas.show()
```

```python
canvas, ax = Canvas.subplots()
ax.pie([30, 45, 25], labels=["A", "B", "C"], autopct="%1.0f%%")
ax.set_title("Shares")
canvas.show()
```

## Regions, arrows, and reference lines

```python
canvas, ax = Canvas.subplots()
ax.plot(x, np.sin(x), color="black")
ax.fill_betweenx([-1, 0, 1], 0.5, [1.0, 1.5, 2.0], alpha=0.2)
ax.axvspan(1.0, 2.0, color="orange", alpha=0.2)
ax.axhspan(-0.25, 0.25, color="steelblue", alpha=0.15)
ax.arrow(2.0, np.sin(2.0), 0.5, 0.3, length_includes_head=True)
ax.axline((0, 0), slope=0.2, linestyle="--", color="crimson")
canvas.show()
```

## Secondary y-axis

```python
twin_canvas, primary = Canvas.subplots()
secondary = twin_canvas.twinx()

primary.plot(x, np.sin(x), color="tab:blue")
secondary.plot(x, 100 * np.cos(x), color="tab:red")
primary.set_xlabel("Time")
primary.set_ylabel("sin(x)", color="tab:blue")
secondary.set_ylabel("100 cos(x)", color="tab:red")
twin_canvas.set_title("Two scales sharing one x-axis")
twin_canvas.show()
```

Secondary y-axes are supported by the Matplotlib and Plotly backends.
The plotext and TikZ backends currently reject a canvas containing
`twinx()` plots instead of silently producing an incorrect figure.

## Scientific field plots

```python
x = np.linspace(-1, 1, 40)
y = np.linspace(-1, 1, 40)
xx, yy = np.meshgrid(x, y)
z = xx**2 + yy**2

canvas, ax = Canvas.subplots()
ax.contour(x, y, z, colors="black")
ax.contourf(x, y, z, alpha=0.5)
ax.pcolormesh(x, y, z, alpha=0.25)
ax.set_title("Scalar field")
canvas.show()
```

For point-density data and matrix-oriented displays:

```python
canvas, ax = Canvas.subplots()
ax.hexbin(xx.ravel(), yy.ravel(), gridsize=12)
ax.matshow(z)
canvas.show()
```

Unstructured triangular data and vector fields are also supported:

```python
points_x = np.array([0.0, 1.0, 0.0, 1.0])
points_y = np.array([0.0, 0.0, 1.0, 1.0])
triangles = [[0, 1, 2], [1, 3, 2]]
values = points_x + points_y

canvas, ax = Canvas.subplots()
ax.quiver(points_x, points_y, np.ones(4), np.ones(4))
ax.triplot(points_x, points_y, triangles=triangles)
ax.tripcolor(points_x, points_y, values, triangles=triangles, alpha=0.3)
ax.tricontour(points_x, points_y, values, triangles=triangles)
canvas.show()
```

## Statistical and event plots

```python
canvas, ax = Canvas.subplots()
ax.stem([0, 1, 2], [1, 3, 2])
ax.stackplot([0, 1, 2], [1, 2, 1], [2, 1, 2], alpha=0.4)
ax.set_title("Discrete and stacked data")
canvas.show()
```

```python
canvas, ax = Canvas.subplots()
ax.boxplot([[1, 2, 3], [2, 4, 5]])
ax.violinplot([[1, 2, 3], [2, 4, 5]])
ax.eventplot([[0.2, 0.5], [1.0, 1.5]])
canvas.show()
```
