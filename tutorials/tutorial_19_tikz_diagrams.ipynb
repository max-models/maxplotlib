# TikZ diagrams

`canvas.add_tikzfigure()` adds a panel that is a
[tikzfigure](https://max-models.github.io/tikzfigure/) diagram instead of a plot:
nodes and the paths between them. The canvas draws it with Matplotlib, next to
ordinary plots, and the diagram itself is native TikZ for LaTeX.

```python
from maxplotlib import Canvas
```

## Nodes and paths

`add_tikzfigure` returns a tikzfigure `TikzFigure`. `add_node(x, y, ...)` places a
node with TikZ options such as `shape`, `fill` and `draw`, and `draw([...])`
connects nodes with a path.

```python
canvas = Canvas(width="12cm", ratio=0.3)
diagram = canvas.add_tikzfigure()

source = diagram.add_node(
    0, 0, label="source", content="source", shape="circle", fill="blue!20"
)
grids = diagram.add_node(
    3, 0, label="grids", content="grids", shape="rectangle", fill="gray!30"
)
dump = diagram.add_node(
    6, 0, label="dump", content="dump", shape="rectangle", fill="red!20"
)
first = diagram.draw([source, grids], color="blue", arrows="->")
second = diagram.draw([grids, dump], color="red", line_width="1.5pt", arrows="->")
```

## With Matplotlib

The canvas draws each node as its shape (`circle` or `rectangle`) in its `fill` and
`draw` colors, with its `content` as text, and each path as a line with its
`color`, `line_width` and `style`. Arrow tips and the other TikZ options are left to
the TikZ version below.

```python
fig, axes = canvas.show()
```

## As TikZ

The diagram is a tikzfigure figure, so printing it gives its TikZ code, with every
option as written, arrow tips included, to `\input` into a LaTeX document:

```python
print(diagram)
```

```python
diagram.show()
```

`canvas.render(backend="tikzfigure")` instead converts the canvas as drawn by
Matplotlib into pgfplots code, like any other canvas, for a figure that combines
the diagram with plots.

## Backends

Diagram panels are drawn by the Matplotlib and tikzfigure backends. The Plotly and
plotext backends refuse a canvas with a diagram panel:

```python
try:
    canvas.render(backend="plotly")
except NotImplementedError as error:
    print(error)
```
